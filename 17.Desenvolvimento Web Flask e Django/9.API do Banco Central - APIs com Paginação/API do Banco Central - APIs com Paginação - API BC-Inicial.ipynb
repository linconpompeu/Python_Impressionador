### API do Banco Central

https://dadosabertos.bcb.gov.br/dataset?res_format=API

```python
import pandas as pd
import requests
```

```python
link = "https://olinda.bcb.gov.br/olinda/servico/mecir_dinheiro_em_circulacao/versao/v1/odata/informacoes_diarias?$top=10000&$orderby=Data%20desc&$format=json"

requisicao = requests.get(link)

informacoes = requisicao.json()
```

```python
tabela = pd.DataFrame(informacoes['value'])

#tabela["Valor"] = tabela["Quantidade"].map("R${:,}".format)



display(tabela)
```

```python
# pegar todas as informacoes com varias requisicoes

tabela_final = pd.DataFrame()
pular_texto = 0

while True:
    link2 = f"https://olinda.bcb.gov.br/olinda/servico/mecir_dinheiro_em_circulacao/versao/v1/odata/informacoes_diarias?$top=10000&$skip={pular_texto}&$orderby=Data%20desc&$format=json"
    requisicao2 = requests.get(link2)
    informacoes2 = requisicao2.json()
    tabela2 = pd.DataFrame(informacoes2['value'])
    if len (informacoes2['value']) < 1:
        break
    tabela_final = pd.concat([tabela_final,tabela2])
    pular_texto += 10000

display(tabela_final)
```
