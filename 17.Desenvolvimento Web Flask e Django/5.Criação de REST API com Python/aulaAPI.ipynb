```python
from flask import Flask

app = Flask(__name__) # cria o site

@app.route("/") # decorator - > diz qual link a função vai rodar
def hello_world(): # funcao
    return {"Hello What?": "Hello, World!"}

@app.route("/pompeu") # decorator - > diz qual link a função vai rodar
def pompeu(): # funcao
    return {"Pergunta": "Qual foi meno"}


app.run() # coloca o site no ar
```
