```python

# passo 1: Entrar no sistema da empresa (link)
import pyautogui
import time

# pyautogui.click -> clicar com o mouse
# pyautogui.write -> escrever um texto
# pyautogui.press -> apertar uma tecla
# pyautogui.hotkey -> combinação de teclas
# pyautogui.pause = 1 -> espera um segundo a cada execução

pyautogui.pause = 2

# Passo a passo do desafio
# Passo 1: abrir o sistema da empresa
pyautogui.press ("win")
pyautogui.write ("chrome")
pyautogui.press("enter")

time.sleep(2)

pyautogui.write ("https://drive.google.com/drive/folders/149xknr9JvrlEnhNWO49zPcw0PW5icxga")
pyautogui.press ("enter")

time.sleep(5) # esperar o carregamento da página
```

```python
# Passo 2: Navegar no sistema para encontrar a base de dados

pyautogui.click(x=372, y=380, clicks=2) #seleciona arquivo
time.sleep(2)
```

```python
# Passo 3: Exportar a base de dados (baixar os arquivos)

pyautogui.click(x=1187, y=382)
time.sleep(2)

pyautogui.click(x=890, y=466)
time.sleep(3)
```

```python
# Passo 4: Calcular os indicadores (Faturamento e quantidade de produtos vendidos)

import pandas as pd

time.sleep(5) # esperar o download do arquivo

caminho = r"C:\Users\lincon.pompeu\Downloads\Vendas - Dez.xlsx"
tabela = pd.read_excel(caminho)

display(tabela)

faturamento = tabela["Valor Final"].sum()

qtde_produtos = tabela["Quantidade"].sum()

print (faturamento)
print (qtde_produtos)
```

```python
import pyperclip

# Passo 5: Enviar as informações por e-mail

pyautogui.pause = 1

#abrir o email
pyautogui.hotkey("ctrl", "t")
pyautogui.press("enter")
time.sleep(2)

#entrar no email
pyautogui.write ("https://mail.google.com/mail/u/1/")
pyautogui.press ("enter")

time.sleep(5)

#clicar no botão escrever
pyautogui.click(x=77, y=230)
time.sleep(2)

# para quem enviar o email
pyautogui.write ("lincon.pompeu+codigo@gmail.com")
time.sleep(3)
pyautogui.press ("tab") # seleciona o email
time.sleep(1)
pyautogui.press ("tab") # passar para o assunto do email
time.sleep(1)

# qual o assunto
pyperclip.copy("Relatório de Vendas")
time.sleep(1)
pyautogui.hotkey("ctrl", "v")
time.sleep(1)
pyautogui.press ("tab") # passar para o corpo do email

# qual o corpo do email
texto = f"""
Prezados,

Segue o relatório de vendas de hoje.

Faturamento: R${faturamento:,.2f}
Quantidade de produtos vendidos: {qtde_produtos:,}

Qualquer duvida, estou à disposição.
"""

pyperclip.copy(texto)
pyautogui.hotkey("ctrl", "v")

# enviar o email

pyautogui.click(x=670, y=1380)

```
