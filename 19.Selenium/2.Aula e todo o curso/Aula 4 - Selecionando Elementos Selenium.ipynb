```python
from selenium.webdriver.common.by import By

campo_email = navegador.find_element(By.ID, "ap-email")

campo_email.send_keys("lincon.pompeu@gmail.com")

campo_entrar = navegador.find_element(By.ID, "ap-submit").click()
```

```python
from selenium.webdriver.common.keys import Keys

navegador.implicitly_wait(5)

campo_email2 = navegador.find_element(By.ID, "email")

campo_email2.send_keys("lincon.pompeu@gmail.com")

campo_email2.send_keys(Keys.ENTER)
```
